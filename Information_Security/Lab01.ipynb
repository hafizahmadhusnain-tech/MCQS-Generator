```python
sum = 0
n = int(input('Enter a number: '))
while n!= 0:
    sum += n
    n = int(input('Enter a number: '))
print(f'The sum is {sum}.')    
```

```python
isPrime = True
i = 2
n = int(input('Enter a number: '))
while i <= n//2:
    if n%i == 0:
        isPrime = False
        break
    i += 1
if isPrime:
    print(f'{n} is prime.')
else:
    print(f'{n} is not prime.')        
```
