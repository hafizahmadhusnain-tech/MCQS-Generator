Simple Stream Cipher

```python
def stream_encrypt(text, key):
    result = ""
    for p, k in zip(text, key):
        p = ord(p)
        k = ord(k)
        encrypted = p ^ k
        result += chr(encrypted)
    return result

def stream_decrypt(ciphertext, key):
    result = ""
    for c, k in zip(ciphertext, key):
        c = ord(c)
        k = ord(k)
        decrypted = c ^ k
        result += chr(decrypted)
    return result
text = "HELLO"
key = "XMCKL"

encrypted = stream_encrypt(text, key)
print("Encrypted:", encrypted)
decrypted = stream_decrypt(encrypted, key)
print("Decrypted:", decrypted)
```

RC4 / KSA 

```python
def rc4(data,key):
    box = list(range(256))
    j = 0
    for i in range(256):
        j = (j + box[i] + ord(key[i%len(key)])) % 256
        box[i], box[j] = box[j],box[i]

    i = 0
    j = 0
    result = ''

    for ch in data:
         i = (i + 1) % 256
         j = ( j + box[i]) % 256

         keystream = box[(box[i] + box[j]) % 256]
         result += chr(ord(ch) ^ keystream)
    return result

data = 'Ahmad'
key = 'abcd'     

x = rc4(data,key)
print(x)
print(rc4(x,key))
```

Vernam Cipher

```python
def vernam(text, key):
    result = ""

    for i, char in enumerate(text):
        result += chr(ord(char) ^ ord(key[i % len(key)]))

    return result


key = "thisismykey12345"

text = 'Ahmad Husnain'

encrypted = vernam(text, key)
print("Encrypted: ", encrypted.encode().hex())

decrypted = vernam(encrypted, key)
print("Decrypted:", decrypted)
```
