Ceaser Cipher

```python
def ceaserCipher(text: str , shift: int) -> str:
    result = ''
    for i in text:
        if i.isalpha():
            base = ord('A') if i.isupper() else ord('a')
            
            result += chr((ord(i) - base + shift) % 26 + base)
        else:
            result+=i    
    return result
encrypt = ceaserCipher('Ahmad Husnain',3)
decrypt = ceaserCipher(encrypt,-3)
print('Cipher ', encrypt)
print('Plain ', decrypt)
```

Vigenre Cipher

```python
def VigenreCipher(text: str, key: str) -> str:
    key_index = 0
    result = ''
    for i in text:
        if i.isalpha():
            result +=  chr(((ord(i.upper())-ord('A') + ord(key[key_index%len(key)].upper())-ord('A'))%26) + ord('A'))
            key_index+=1 
        else:
            result+=i
           
    return result

def DecryptionText(cipher: str ,key: str) -> str:
    result = ''
    key_index = 0

    for i in cipher:
        if i.isalpha():
            result += chr((((ord(i.upper())-ord('A')) - (ord(key[key_index%len(key)].upper())-ord('A'))) %26) + ord('A'))
            key_index +=1
        else:
            result+=i    
    return result

cipher = VigenreCipher('Ahmad','key')
plain = DecryptionText(cipher,'key')
print('Cipher: ', cipher)
print('Plain: ', plain)
```

One-Time-Pad Cipher

```python
def opt_cipher(text: str, key: str) -> str:
    if len(text) != len(key):
        raise ValueError('OTP key must same length as plain text')
    result = ''
    for i,j in zip(text,key):
        c =  ord(i.upper()) - ord('A')
        k = ord(j.upper()) - ord('A')
        r = (c + k ) % 26
        result += chr(r + ord('A'))
    return result
cipher = opt_cipher('Ahmad','Hasan')

def decrypt(cipher, key):
    if len(cipher) != len(key):
        raise ValueError('OTP key must same length as plain text')

    result = ''
    for i,j in zip(cipher,key):
        p = ord(i.upper())- ord('A')
        k = ord(j.upper()) - ord('A')
        r = (p-k)%26
        result += chr(r+ord('A'))
    return result
plain = decrypt(cipher,'Hasan')

print(cipher)
print(plain)
```
