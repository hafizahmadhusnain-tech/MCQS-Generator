```python
import cv2
import sys

img = cv2.imread('1.png')

newImg = cv2.resize(img,(600,400))

cv2.imshow('My IMG',newImg)
cv2.waitKey(0)
cv2.imwrite('resized_img.jpg', newImg)
cv2.destroyAllWindows()
```

```python
cropped_img = newImg[50:350 , 100:400]
cv2.imshow('My IMG',newImg)
cv2.imshow('h',cropped_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gray_img = cv2.cvtColor(newImg , cv2.COLOR_BGR2GRAY)
cv2.imshow('h',gray_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
blurImg = cv2.GaussianBlur(newImg,(19,19),0)
cv2.imshow('h',blurImg)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
edge_detect1 = cv2.Canny(gray_img , 100,200)
edge_detect = cv2.Canny(gray_img , 200,100)

cv2.imshow('h',edge_detect1)


cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.line(newImg,(70,70),(250,200),(0,255,0),2)
cv2.rectangle(newImg,(70,70),(250,200),(0,255,0), 2)
cv2.circle(newImg,(300,300),50,(255,0,0),2)

cv2.imshow('h2',newImg)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
font = cv2.FONT_HERSHEY_DUPLEX
cv2.putText(newImg.copy(),'Ahmad Husnain',(300,50),font,fontScale=1,color=(255,255,255))
cv2.imshow('h2',newImg)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
