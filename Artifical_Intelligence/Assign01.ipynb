```python
from collections import deque

graph1 = {
    10:[7,15],
    7:[4,9],
    15:[20],
    4:[],
    9:[],
    20:[]
}

start = list(graph1.keys())[0]
queue = deque()
visited = set()

queue.append(start)
while queue:
    node = queue.popleft()
    if node not in visited:
        print(node , end=' ')
        visited.add(node)
    for i in reversed(graph1[node]):
        if i not in visited:
            queue.append(i)


```

Q No.2
DFS

```python
tree = {
    1: [2, 3],
    2: [4,5],
    3: [6,None],
    5: [7,None],
    6:[8,None]
}
def maxDepth(node):
    if node is None:
        return 0

    left, right = tree.get(node, [None, None])

    left_depth = maxDepth(left)
    right_depth = maxDepth(right)

    return 1 + max(left_depth, right_depth)
root = list(tree.keys())[0]
print(f'Max Depth of tree is: {maxDepth(root)}')
```

Q No.3
Words Finding solution from grid

```python
def exist(board, word):
    rows,cols= len(board),len(board[0])
    def dfs(row, col, index):
        if index == len(word):
            return True
        if row < 0 or row >= rows or col < 0 or col >= cols:
            return False
        if board[row][col] != word[index]:
            return False
        temp = board[row][col]
        board[row][col] = "#"
        found = (dfs(row + 1, col, index + 1) or
                 dfs(row - 1, col, index + 1) or
                 dfs(row, col + 1, index + 1) or
                 dfs(row, col - 1, index + 1))
        board[row][col] = temp
        return found
    for row in range(rows):
        for col in range(cols):
            if dfs(row, col, 0):
                return True
    return False
board = [['A', 'B', 'C', 'E'],
         ['H', 'M', 'C', 'S'],
         ['A', 'A', 'D', 'E']]
word = "AHMAD"
print(f'Given word "{word}" Found!') if exist(board,word) else print(f'Given word "{word}" not Found!')
```
