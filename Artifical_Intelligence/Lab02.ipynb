```python
def chatbot():
    faqs = {
        "hello": "Hi there! How can I help you today?",
        "hi": "Hello! What can I do for you?",
        "how are you": "I'm just a bunch of code, but I'm running fine!",
        "what is your name": "I'm PyBot, your friendly FAQ assistant.",
        "what can you do": "I can answer basic questions from my FAQ list.",
        "what is python": "Python is a popular high-level programming language.",
        "who created python": "Python was created by Guido van Rossum in 1991.",
        "what is a variable": "A variable is a named container for storing data.",
        "what is a function": "A function is a reusable block of code that performs a task.",
        "help": "Try asking about Python, variables, functions, or say 'bye' to exit.",
        "thanks": "You're welcome!",
        "thank you": "Anytime! Happy coding.",
    }

    exit_words = {"quit", "bye", "exit", "q"}

    print("🤖 PyBot: Hello! Ask me something (type 'bye' or 'quit' to exit).")

    while True:
        user_input = input("You: ").strip().lower()

        if not user_input:
            print("🤖 PyBot: Please type something.")
            continue

        # Check for exit
        if user_input in exit_words:
            print("🤖 PyBot: Goodbye! Have a great day. 👋")
            break

        # Look up FAQ
        if user_input in faqs:
            print(f"🤖 PyBot: {faqs[user_input]}")
        else:
            print("🤖 PyBot: Sorry, I don't know that one. Type 'help' to see options.")


if __name__ == "__main__":
    chatbot()
```

```python
import os
os.environ["HF_HUB_DISABLE_SYMLINKS_WARNING"] = "1"

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

MODEL_NAME = 'facebook/opt-125m'

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model     = AutoModelForCausalLM.from_pretrained(MODEL_NAME)

query = 'What is Apple fruit ?'

# 1. Encode
encoded = tokenizer(query, return_tensors='pt', padding=True)
input_ids      = encoded["input_ids"]
attention_mask = encoded["attention_mask"]

output_ids = model.generate(
    input_ids=input_ids,               
    attention_mask=attention_mask,     
    max_new_tokens=100,
    temperature=0.2,
    do_sample=True,
    pad_token_id=tokenizer.pad_token_id,
)

# 3. Decode
text = tokenizer.decode(output_ids[0], skip_special_tokens=True)
print(text)
```

```python
fruits = ['Apple', 'Apricoate', 'Banana', 'Mango', 'melon']


def compare_strings(a, b):
    """Return:
       -1 if a < b
        0 if a == b
        1 if a > b
    """

    i = 0

    while i < len(a) and i < len(b):

        if a[i] < b[i]:
            return -1

        if a[i] > b[i]:
            return 1

        i += 1

    if len(a) < len(b):
        return -1
    elif len(a) > len(b):
        return 1
    else:
        return 0


def binary_search(fruits, target):

    low = 0
    high = len(fruits) - 1

    while low <= high:

        mid = (low + high) // 2

        result = compare_strings(fruits[mid], target)

        if result == 0:
            return mid

        elif result < 0:
            low = mid + 1
        else:
            high = mid - 1
    return -1

print(binary_search(fruits, 'Banana'))
```
