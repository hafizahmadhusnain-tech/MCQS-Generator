Graphs of Matplotlib

```python
import matplotlib.pyplot as plt
import pandas as pd

dates = pd.date_range("2022-01-01", "2022-12-31", periods=12)

open_prices = [
    36300, 35500, 34800, 34500,
    33500, 32500, 31500, 32500,
    33500, 31000, 33000, 33500
]

close_prices = [
    36500, 35200, 35000, 34200,
    33800, 32000, 31800, 33000,
    34000, 30500, 33500, 33200
]

plt.plot(dates, open_prices, label="Open")
plt.plot(dates, close_prices, label="Close")

plt.title("DJIA Open and Close Prices")
plt.xlabel("Date")
plt.ylabel("Price")

plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

years = [2015, 2016, 2017, 2018, 2019]

IT = [12, 30, 1, 8, 22]
ECE = [28, 6, 16, 5, 10]
CSE = [29, 3, 24, 25, 17]

x = np.arange(len(years))

width = 0.25

plt.bar(x - width, IT, width, label="IT")
plt.bar(x, ECE, width, label="ECE")
plt.bar(x + width, CSE, width, label="CSE")

plt.xlabel("Branch")
plt.ylabel("Students passed")

plt.xticks(x, years)

plt.legend()

plt.show()
```
