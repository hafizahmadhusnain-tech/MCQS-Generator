Water Container Problem

```python
def water_container(height: list[int]) -> int:
    left = 0
    right = len(height) - 1
    max_water = 0

    while left < right:
        width = right - left
        current_height = min(height[left], height[right])
        current_area = width * current_height
        max_water = max(max_water , current_area)
        if height[left] < height[right]:
            left+=1
        else:
            right-=1
    return max_water
x = water_container([1,8,6,2,5,7])
x                
```

INT TO ROMAN

```python
def int_to_roman(num: int) -> str:
    res = ''
    roman_map = [
        (1000,'M'),(900,'CM'),(500,'D'),(400,'CD'),
        (100,'C'),(90,'XC'),(50,'L'),(40,'XL'),
        (10,'X'),(9,'IX'),(5,'V'),(4,'IV'),(1,'I')
    ]
    for val, syb in roman_map:
        while num>=val:
            res+=syb
            num-=val
            
    return res

print(int_to_roman(7))
```
