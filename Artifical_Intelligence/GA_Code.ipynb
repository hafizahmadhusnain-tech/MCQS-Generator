```python
import random as rd 

def fitness(chromosomes:list) -> int:
    conflits = 0
    n = len(chromosomes)

    for i in range(n):
        for j in range(i+1,n):
            if abs(chromosomes[i]-chromosomes[j]) == abs(i-j):
                conflits+=1
    total_pair = n * (n-1) // 2            
    return  total_pair - conflits           
```

```python
def selection(population):
    score = [fitness(ch) + 1 for ch in population]

    return rd.choices(
        population, weights=score, k=1
    )[0]
```

```python
def crossover(parent1,parent2):
    n = len(parent1)

    a,b = sorted(rd.sample(range(n),2))
    child = [None] * n

    child[a:b] = parent1[a:b]

    values = (
        parent2[b:] + parent2[:b]
    )

    values = [x for x in values if x not in child]

    position = list(range(b,n)) + list(range(a))

    for pos, val in zip(position,values):
        child[pos] = val

    return child

x = crossover([2,3,4,1],[1,3,2,4])
x
```

```python
def mutation(chromosome: list ) -> list:
    chromosome = chromosome.copy()

    if rd.random() < 0.2:
        i,j = rd.sample(range(len(chromosome)),2)
        chromosome[i],chromosome[j] = (chromosome[j],chromosome[i])

    return chromosome    
```

```python
def genetic_algorithim(n):
    max_population = 50
    max_gen = 1000
    max_fitness = n*(n-1)//2

    population = [
        rd.sample(range(1,n+1),n) for _ in range(max_population)
    ]

    for gen in range(max_gen):

        population.sort(
            key=fitness,
            reverse=True
        )

        best = population[0]

        if fitness(best) == max_fitness:
            print(f"\n{n}-Queens Solution Found!")
            print("Solution:", best)
            print("Fitness:", fitness(best))
            print("Generation:", gen)
            return best

        new_population = population[:2]

        while len(new_population) < max_population:
            p1 = selection(population)
            p2 = selection(population)

            child = crossover(p1,p2)

            child = mutation(child)

            new_population.append(child)
        population = new_population

    print(f"No solution found for {n}-Queens.")
    return None

rd.seed(42)
n = 8
solution = genetic_algorithim(n)

if solution:
    for i in range(1,n+1):
        for j in range(n):
            if solution[j] == i:
                 print("Q", end=" ")
            else:
                 print(".", end=" ")  
        print()            
```
